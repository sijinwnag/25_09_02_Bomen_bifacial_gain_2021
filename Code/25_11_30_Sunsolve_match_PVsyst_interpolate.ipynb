# 1. Data loading 

## 1.1. SunSolve Yield data loading

```python
# Load the SunSolve yield data using interpolation
print("=== LOADING ALBEDO SCENARIO FILES ===")

# Load the albedo scenario files (conditionally load veg and soil if available)
zero_albedo_df = pd.read_csv(zero_albedo_file)
const_albedo_df = pd.read_csv(const_albedo_file)

print(f"Loaded zero_albedo_df: {len(zero_albedo_df)} rows")
print(f"Loaded const_albedo_df: {len(const_albedo_df)} rows")

# Conditionally load veg and soil albedo files if paths are provided
if veg_albedo_file is not None:
    veg_albedo_df = pd.read_csv(veg_albedo_file)
    print(f"Loaded veg_albedo_df: {len(veg_albedo_df)} rows")
else:
    veg_albedo_df = None
    print("Veg albedo file not available - will use const gradient")

if soil_albedo_file is not None:
    soil_albedo_df = pd.read_csv(soil_albedo_file)
    print(f"Loaded soil_albedo_df: {len(soil_albedo_df)} rows")
else:
    soil_albedo_df = None
    print("Soil albedo file not available - will use const gradient")

# Extract the power columns
zero_albedo_power = zero_albedo_df['Power [unit-system] (W)']
const_albedo_power = const_albedo_df['Power [unit-system] (W)']

# Extract veg and soil power if available, otherwise use zero as baseline
if veg_albedo_df is not None:
    veg_albedo_power = veg_albedo_df['Power [unit-system] (W)']
else:
    veg_albedo_power = zero_albedo_power

if soil_albedo_df is not None:
    soil_albedo_power = soil_albedo_df['Power [unit-system] (W)']
else:
    soil_albedo_power = zero_albedo_power

# Calculate the gradient of power with respect to albedo in the unit of W/%
print("\n=== CALCULATING POWER GRADIENTS ===")
const_albedo_gradient = (const_albedo_power - zero_albedo_power) / 30

# Calculate veg and soil gradients if files are available, otherwise use const gradient
if veg_albedo_df is not None:
    veg_albedo_gradient = (veg_albedo_power - zero_albedo_power) / 24.391
    print(f"veg_albedo_gradient stats: mean={veg_albedo_gradient.mean():.2f}, std={veg_albedo_gradient.std():.2f}")
else:
    veg_albedo_gradient = const_albedo_gradient
    print("Using const_albedo_gradient for veg spectrum")

if soil_albedo_df is not None:
    soil_albedo_gradient = (soil_albedo_power - zero_albedo_power) / 15.351
    print(f"soil_albedo_gradient stats: mean={soil_albedo_gradient.mean():.2f}, std={soil_albedo_gradient.std():.2f}")
else:
    soil_albedo_gradient = const_albedo_gradient
    print("Using const_albedo_gradient for soil spectrum")

print(f"const_albedo_gradient stats: mean={const_albedo_gradient.mean():.2f}, std={const_albedo_gradient.std():.2f}")

# Create an empty list to store monthly dataframes
monthly_dfs = []

print("\n=== INTERPOLATING MONTHLY DATA ===")
# For each month, interpolate the results
for month in range(12):
    print(f"Processing month {month+1}...")
    
    # Extract this part of the data from zero_albedo_df
    month_zero_df = zero_albedo_df[start_day_index_month[month]:start_day_index_month[month+1]].copy()
    
    # Look up the albedo from the monthly_albedo list
    albedo = monthly_albedo[month]
    
    # Look up the ground spectrum from the ground_spectrum_list
    ground_spectrum = ground_spectrum_list[month]
    
    # Load the corresponding albedo gradient
    if ground_spectrum == 'veg':
        albedo_gradient = veg_albedo_gradient
    elif ground_spectrum == 'soil':
        albedo_gradient = soil_albedo_gradient
    else:
        albedo_gradient = const_albedo_gradient
    
    # Select the part of gradient that is corresponding to this month
    albedo_gradient_month = albedo_gradient[start_day_index_month[month]:start_day_index_month[month+1]]
    
    # Interpolate the power at the albedo index
    month_zero_df['Power [unit-system] (W)'] = month_zero_df['Power [unit-system] (W)'] + albedo_gradient_month.values * albedo
    
    # Append to list
    monthly_dfs.append(month_zero_df)

# Concatenate all monthly dataframes into one
sunsolve_df = pd.concat(monthly_dfs, ignore_index=True)
print(f"\n=== INTERPOLATION COMPLETE ===")
print(f"Combined dataframe shape: {sunsolve_df.shape}")
print(f"Columns in sunsolve_df: {sunsolve_df.columns.tolist()}")

# Create a timestamp column from the Day of year, Hour, and Minute columns
year = 2021
base_date = pd.Timestamp(f"{year}-01-01")  # January 1st of the specified year
sunsolve_df['Timestamp'] = sunsolve_df.apply(
    lambda row: base_date + pd.Timedelta(days=int(row['Day of year'])-1) + 
                pd.Timedelta(hours=int(row['Hour'])) + 
                pd.Timedelta(minutes=int(row['Minute'])),
    axis=1
)

# Sort by timestamp
sunsolve_df.sort_values('Timestamp', inplace=True)

# Define a scaling factor for the SunSolve df power
sunsolve_scaling_1 = 1e-3

# Convert power from W to MW
sunsolve_df['Inverter power (MW)'] = sunsolve_df['Power [unit-system] (W)'] * sunsolve_scaling_1

# Set the timestamp as the index for easier resampling
sunsolve_df.set_index('Timestamp', inplace=True)

# Now resample to daily data using the index
# For hourly data converted to daily energy, we need to multiply by the time period
# The 'Period (h)' column contains the duration of each data point in hours
sunsolve_df['Hourly energy (MWh)'] = sunsolve_df['Inverter power (MW)'] * sunsolve_df['Period (h)']

# Identify all columns containing "(W)" - these are power measurements
power_columns = [col for col in sunsolve_df.columns if "(W)" in col]

# Create energy columns for each power column
for col in power_columns:
    # Create a new column name by replacing "(W)" with "(Wh)"
    energy_col_name = col.replace("(W)", "(Wh)")
    
    # Convert power to energy by multiplying by the time period
    sunsolve_df[energy_col_name] = sunsolve_df[col] * sunsolve_df['Period (h)']

# Resample all energy columns to daily totals
daily_energy_df = sunsolve_df[[col for col in sunsolve_df.columns if "(Wh)" in col]].resample('D').sum()

# Convert from Wh to MWh for consistency with your other calculations
for col in daily_energy_df.columns:
    daily_energy_df[col] = daily_energy_df[col] * 1e-6  # Convert Wh to MWh
    new_col_name = col.replace("(Wh)", "(MWh/day)")
    daily_energy_df.rename(columns={col: new_col_name}, inplace=True)

# Display the results
print("\nDaily energy values from power measurements:")
print(daily_energy_df.head())

# Initialize sunsolve_daily with the hourly energy resampled to daily
sunsolve_daily = sunsolve_df['Hourly energy (MWh)'].resample('D', closed='left', label='left').sum().to_frame(name='Inverter energy (MWh/day)')

# Merge with the other energy columns
sunsolve_daily = pd.concat([sunsolve_daily, daily_energy_df], axis=1)

# Print the columns of the sunsolve_daily
print("\nColumns in daily resampled SunSolve data:")
print(sunsolve_daily.columns)
```

# 3. Optimisation

```python
print("=== OPTIMIZING SUNSOLVE SCALING FACTOR ===")
print("Target MBE tolerance: 1e-13 MWh/day")
print("WARNING: Running with unlimited iterations and no factor boundaries.")
print("         Press Ctrl+C to interrupt if the search takes too long.")

DATE_RANGE = ("2021-01-01", "2021-12-31")  # set to None for full series
# DATE_RANGE = None

sunsolve_daily_full = (
    sunsolve_df['Power [unit-system] (W)']
    * sunsolve_df['Period (h)']
    * 1e-6
).resample('D').sum().to_frame('SunSolve energy (MWh/day)')


if DATE_RANGE is not None:
    start_date, end_date = map(pd.Timestamp, DATE_RANGE)
    sunsolve_filtered = sunsolve_daily_full.loc[start_date:end_date]
    pvsyst_filtered = PVsyst_daily.loc[start_date:end_date]
    window_label = f"{start_date.date()} → {end_date.date()}"
else:
    sunsolve_filtered = sunsolve_df
    pvsyst_filtered = PVsyst_daily
    window_label = "full period"

# def calculate_mbe(scale_factor):
#     """Mean Bias Error restricted to the filtered date window."""
#     scaled_energy = (
#         sunsolve_filtered['Power [unit-system] (W)']
#         * scale_factor
#         * sunsolve_filtered['Period (h)']
#         * 1e-6
#     )
#     sunsolve_daily = scaled_energy.resample('D').sum().to_frame('Scaled Inverter energy (MWh/day)')
#     merged = pd.merge(sunsolve_daily, pvsyst_filtered, left_index=True, right_index=True, how='inner')
#     merged.dropna(inplace=True)
#     return np.mean(merged['Scaled Inverter energy (MWh/day)'] - merged['PVsyst energy (MWh/day)'])

def calculate_mbe(scale_factor):                                                                                              
   # Use pre-calculated daily data instead of raw hourly filtering                                                           
    scaled_daily_energy = sunsolve_daily_full * scale_factor                                                                  
    sunsolve_daily_scaled = scaled_daily_energy.loc[start_date:end_date]                                                      
                                                                                                                               
    merged = pd.merge(sunsolve_daily_scaled, pvsyst_filtered,                                                                 
                      left_index=True, right_index=True, how='inner')                                                          
    merged.dropna(inplace=True)                                                                                              

    return np.mean(merged['SunSolve energy (MWh/day)'] - merged['PVsyst energy (MWh/day)']) 

initial_mbe = calculate_mbe(sunsolve_scaling_1)
if initial_mbe > 0:
    min_factor = sunsolve_scaling_1 * 0.01
    max_factor = sunsolve_scaling_1
else:
    min_factor = sunsolve_scaling_1
    max_factor = sunsolve_scaling_1 * 100

def find_optimal_scaling_factor(min_factor=min_factor, max_factor=max_factor):
    print(f"\nOptimising scale for {window_label}")
    print(f"Initial search range: [{min_factor:.10f}, {max_factor:.10f}]")
    print(f"Initial MBE: {initial_mbe:.10e}")
    iterations = 0
    best_factor = None
    best_mbe = float('inf')

    while True:
        iterations += 1
        mid_factor = (min_factor + max_factor) / 2
        mbe = calculate_mbe(mid_factor)

        if abs(mbe) < abs(best_mbe):
            best_factor, best_mbe = mid_factor, mbe

        if iterations % 10 == 0 or iterations < 10:
            print(f"Iteration {iterations}: Factor = {mid_factor:.10f}, MBE = {mbe:.10e}")

        if abs(mbe) < 1e-13:
            print("✓ Converged! Found factor with MBE below tolerance.")
            return mid_factor, mbe, iterations

        if mbe > 0:
            max_factor = mid_factor
        else:
            min_factor = mid_factor

        if iterations % 100 == 0:
            if abs(min_factor - mid_factor) < 1e-10:
                new_min = min_factor * 0.1
                print(f"Expanding search range downward: {min_factor:.10f} -> {new_min:.10f}")
                min_factor = new_min
            if abs(max_factor - mid_factor) < 1e-10:
                new_max = max_factor * 10
                print(f"Expanding search range upward: {max_factor:.10f} -> {new_max:.10f}")
                max_factor = new_max

optimal_factor_filtered, final_mbe, iterations = find_optimal_scaling_factor()

print("\n=== OPTIMIZATION RESULTS ===")
print(f"Date window: {window_label}")
print(f"Original scaling factor: {sunsolve_scaling_1:.10f}")
print(f"Optimal scaling factor (filtered): {optimal_factor_filtered:.10f}")
print(f"Final MBE: {final_mbe:.12e} MWh/day")
print(f"Found in {iterations} iterations")

print("\n=== APPLYING OPTIMAL SCALING FACTOR (FILTERED WINDOW) ===")
# Apply optimal scaling to pre-calculated daily data
sunsolve_daily_optimized_filtered = (
    sunsolve_filtered * optimal_factor_filtered
).rename(columns={'SunSolve energy (MWh/day)': 'Optimized Inverter energy (MWh/day)'})


merged_optimized = pd.merge(
    sunsolve_daily_optimized_filtered, pvsyst_filtered, left_index=True, right_index=True, how='inner'
)
merged_optimized.dropna(inplace=True)

optimized_rmse = np.sqrt(mean_squared_error(
    merged_optimized['PVsyst energy (MWh/day)'],
    merged_optimized['Optimized Inverter energy (MWh/day)']
))
optimized_mbe = np.mean(
    merged_optimized['Optimized Inverter energy (MWh/day)'] - merged_optimized['PVsyst energy (MWh/day)']
)
optimized_nrmse = optimized_rmse / (
    merged_optimized['PVsyst energy (MWh/day)'].max() - merged_optimized['PVsyst energy (MWh/day)'].min()
)

print("\n=== METRICS WITH OPTIMAL SCALING (FILTERED) ===")
print(f"RMSE: {optimized_rmse:.4f} MWh/day")
print(f"MBE: {optimized_mbe:.4e} MWh/day")
print(f"NRMSE: {optimized_nrmse:.4f}")

plt.figure(figsize=long_hoz_figsize)
plt.plot(merged_optimized.index, merged_optimized['Optimized Inverter energy (MWh/day)'],
         label=f'SunSolve Optimized (scale: {optimal_factor_filtered:.6f})', color='green')
plt.plot(merged_optimized.index, merged_optimized['PVsyst energy (MWh/day)'],
         label='PVsyst Array Energy', color='orange')
plt.xlabel('Date', fontsize=axis_label_size)
plt.ylabel('Daily Energy (MWh/day)', fontsize=axis_label_size)
plt.title(f'Daily Energy Comparison — {window_label}\nMBE: {optimized_mbe:.2e} MWh/day, RMSE: {optimized_rmse:.4f} MWh/day',
          fontsize=title_size)
plt.legend(fontsize=text_size)
plt.tight_layout()
plt.show()

plt.figure(figsize=long_hoz_figsize)
plt.plot(merged_optimized.index,
         merged_optimized['Optimized Inverter energy (MWh/day)'] - merged_optimized['PVsyst energy (MWh/day)'],
         color='black')
plt.axhline(y=0, color='r', linestyle='-', alpha=0.3)
plt.xlabel('Date', fontsize=axis_label_size)
plt.ylabel('Difference (MWh/day)', fontsize=axis_label_size)
plt.title(f'Difference between Optimized SunSolve and PVsyst ({window_label})', fontsize=title_size)
plt.tight_layout()
plt.show()
```
